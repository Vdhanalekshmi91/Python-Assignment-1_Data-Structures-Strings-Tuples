## String Concatenation:

```python
string_1="Hello "

string_2=input(" enter your name ")

print(string_1 + string_2)
```

```python
ext_string=string_1 + string_2

string_3=", welcome to python programming" 

final_string=ext_string + string_3

print(final_string)
```

## String Slicing and Indexing:

```python
text=final_string

print(text)
```

```python
print(text[0])
print(text[-1])
print(text[:5])
print(text[-11:])
print(text[::-1])
print(text[33:39])
```
